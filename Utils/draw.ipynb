```python

import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# 示例数据（请替换为你自己的数据）
codet5_compile_errors = [0.5, 0.7, 1.0, 1.0, 0.4, 0.9, 1.0, 0.4, 0.9, 0.8, 0.9, 1.0, 0.8, 1.0, 1.0, 1.0, 0.6, 0.9, 1.0, 0.5, 0.5, 1.0, 0.6, 0.9, 1.0, 1.0, 0.9, 0.5, 0.9, 0.7, 1.0, 0.9, 0.5, 1.0, 1.0, 0.5, 0.4, 0.4, 0.4, 1.0, 0.5, 0.4, 1.0, 0.9, 0.3, 0.6, 1.0, 0.2, 0.3, 0.3, 0.8, 1.0, 0.5, 0.8, 1.0, 0.7, 0.9, 0.6]
prooft5_compile_errors = [0.05, 0.07, 0.08, 0.06, 0.09, 0.07, 0.06, 0.05, 0.04, 0.03]

# 检查长度是否一致
assert len(codet5_compile_errors) == len(prooft5_compile_errors)

n = len(codet5_compile_errors)
problems = [f"P{i+1}" for i in range(n)]

# 设置风格
sns.set(style="whitegrid")
plt.figure(figsize=(14, 7))

# 设置柱子宽度
bar_width = 0.4
index = np.arange(n)

# 绘制柱状图
plt.bar(index, codet5_compile_errors, bar_width, label='Codet5', color='#4E79A7')
plt.bar(index + bar_width, prooft5_compile_errors, bar_width, label='ProoF-T5', color='#F28E2B')

# 添加标题和标签
plt.title('Compile Error Rate per Problem: Codet5 vs ProoF-T5', fontsize=16)
plt.xlabel('Problem', fontsize=12)
plt.ylabel('Compile Error Rate', fontsize=12)
plt.xticks(index + bar_width / 2, problems)
plt.ylim(0, max(max(codet5_compile_errors), max(prooft5_compile_errors)) + 0.05)

# 图例和布局优化
plt.legend(loc='upper right', fontsize=12)
plt.tight_layout()

# 显示图像
plt.show()
```
